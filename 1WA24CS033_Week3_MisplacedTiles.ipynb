```python
import heapq

def manhattan_distance(state, goal):
    distance = 0

    for i in range(3):
        for j in range(3):
            tile = state[i][j]

            if tile != 0:
                for x in range(3):
                    for y in range(3):
                        if goal[x][y] == tile:
                            distance += abs(i - x) + abs(j - y)

    return distance


def get_neighbors(state):
    neighbors = []
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                row, col = i, j
                break
        else:
            continue
        break
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_state = [r[:] for r in state]
            new_state[row][col], new_state[new_row][new_col] = (
                new_state[new_row][new_col],
                new_state[row][col]
            )

            neighbors.append(new_state)

    return neighbors


def a_star(initial, goal):
    open_list = []
    heapq.heappush(
        open_list,
        (manhattan_distance(initial, goal), 0, initial, [initial])
    )

    closed = set()
    best_g = {tuple(map(tuple, initial)): 0}

    while open_list:
        f, g, current, path = heapq.heappop(open_list)
        current_key = tuple(map(tuple, current))

        if current_key in closed:
            continue
        if current == goal:
            return path

        closed.add(current_key)

        for new_state in get_neighbors(current):
            new_key = tuple(map(tuple, new_state))
            new_g = g + 1

            if new_key not in closed and (
                new_key not in best_g or new_g < best_g[new_key]
            ):
                best_g[new_key] = new_g
                h = manhattan_distance(new_state, goal)
                new_f = new_g + h

                heapq.heappush(
                    open_list,
                    (new_f, new_g, new_state, path + [new_state])
                )

    return None


def print_puzzle(state):
    for row in state:
        print(" ".join("_" if tile == 0 else str(tile)
                       for tile in row))
    print()
initial = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]
goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]
solution = a_star(initial, goal)

if solution:
    print("Goal state reached!")
    print("Number of moves:", len(solution) - 1)
    print("Solution path:\n")

    for step, state in enumerate(solution):
        print("Step", step)
        print_puzzle(state)
else:
    print("No solution found.")
```

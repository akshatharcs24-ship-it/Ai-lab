```python

import random
import math

# Fitness function
def fitness(x):
    return x ** 2

# Parameters
NESTS = 10
ITERATIONS = 50
PA = 0.25

# Initialize nests
nests = [random.uniform(-10, 10)
         for _ in range(NESTS)]

best = max(nests, key=fitness)

for iteration in range(ITERATIONS):

    for i in range(NESTS):

        # Generate new solution using Levy-like step
        step = random.gauss(0, 1)
        new_solution = nests[i] + step

        # Boundary checking
        new_solution = max(-10, min(10, new_solution))

        # Select random nest
        j = random.randrange(NESTS)

        # Replace if better
        if fitness(new_solution) > fitness(nests[j]):
            nests[j] = new_solution

    # Abandon worst nests
    num_abandon = int(PA * NESTS)

    nests.sort(key=fitness, reverse=True)

    for i in range(NESTS - num_abandon, NESTS):
        nests[i] = random.uniform(-10, 10)

    # Update best solution
    current_best = max(nests, key=fitness)

    if fitness(current_best) > fitness(best):
        best = current_best

# Output
print("Best Solution:", best)
print("Maximum Fitness:", fitness(best))
```

```python
def forward_chaining(rules, facts, query):
    # Keep track of known facts in a set
    known_facts = set(facts)
    added_new = True

    while added_new:
        added_new = False
        for premises, conclusion in rules:
            # If all premises are known and the conclusion is not yet known
            if all(p in known_facts for p in premises):
                if conclusion not in known_facts:
                    known_facts.add(conclusion)
                    added_new = True
                    if conclusion == query:
                        return True

    return query in known_facts

# Separate rules and facts
rules = {
    (("A",), "B"),
    (("B",), "C"),
    (("C",), "D"),
    (("A", "D"), "E")
}
facts = {"A"}
query = "E"

result = forward_chaining(rules, facts, query)

if result:
    print("Query is TRUE")
else:
    print("Query is FALSE")
```

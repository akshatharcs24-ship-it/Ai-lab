```python

from itertools import product

# Truth table enumeration for propositional logic

def implies(a, b):
    return (not a) or b

def iff(a, b):
    return a == b

def check_KB(P, Q, R):
    s1 = iff(P, Q)          # P <-> Q
    s2 = Q or R             # Q OR R
    s3 = implies(not P, R)  # NOT P -> R

    return s1 and s2 and s3

print("P Q R | P<->Q | Q OR R | NOT P -> R | KB")

models = []

for P, Q, R in product([True, False], repeat=3):

    s1 = iff(P, Q)
    s2 = Q or R
    s3 = implies(not P, R)

    KB = s1 and s2 and s3

    print(int(P), int(Q), int(R), " | ",
          int(s1), "    | ",
          int(s2), "     | ",
          int(s3), "       | ",
          int(KB))

    if KB:
        models.append((P, Q, R))

print("\nModels of KB:", models)

# Check entailment of a query
def entails(query):
    for P, Q, R in product([True, False], repeat=3):
        if check_KB(P, Q, R) and not query(P, Q, R):
            return False
    return True

print("\nKB entails Q:", entails(lambda P,Q,R: Q))

print("KB entails P -> Q:",
      entails(lambda P,Q,R: implies(P,Q)))

print("KB entails R -> Q:",
      entails(lambda P,Q,R: implies(R,Q)))
```
